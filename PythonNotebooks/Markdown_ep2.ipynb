## Bağlantılar (Links)
Metinleri tıklanabilir hale getirerek başka sayfalara, dokümanlara veya dosyalara yönlendirmemizi sağlar.

#### Satır İçi Bağlantı (Inline Link) - En yaygın Yöntemdir.
Bu bir [Google](https://google.com) Bağlantısıdır.

#### Satır İçi Opsiyonel Başlık (Hover yapınca açıklaması görülür.)
Bu da [Github](https://github.com "Github Anasayfa") bağlantısıdır.

#### Referans Bağlantısı (Reference Link) - Temiz Kod için
Bu metinde [Python][1] ve [Markdown][2] kullanılmıştır.

... (Dokümanın altı veya herhangi bir yeri) ...

[1]: https://python.org
[2]: htpps://daringfireball.net/projects/markdown/

## Görseller (Images/Resizing)
Link yapısına çok benzer. Tek farkı başındaki ünlem işaretidir.

Standart format şöyledir. -> ![Alt Metin](Görsel URL'si)

! -> Bu bir görseldir, demek
[] -> Görsel Yüklenemezse görünecek metin, erişilebilirlik için önemlidir.
() -> Resmin dosya yolu veya internet sitesi

### Boyutlandırma (Resizing) Sorunu
- Standart markdownda ve GitHub Flavored Markdown'da görsel boyutlandırmak için (width="500" gibi) doğrudan bir syntax yoktur.
- Mühendislik dökümantasyonlarında boyutlandırma gerektiğinde saf html kullanılır.

### Alıntılar (Quotes)
- Bir metni başka bir kaynaktan aldığımızı belirtmek ve
- Okuyucunun dikkatini çeken özel not alanları için kullanılır.
- Epostalardaki yanıt formatına benzer.

> Bu bir alıntı bloğudur.
> Markdown, sadeliği ile güçlüdür.
>
> Araya boş ">" koyarak alıntı içi yeni paragraf da oluşturabiliriz.


##### İç içe (nested)
* ">" sayısını arttırarak alıntı içinde alıntı yapabiliriz.

> Ana alıntı seviyesi 1.
>> İçerideki alıntı seviyesi 2.
>
>  Tekrar seviye 1.

##### Alıntı bloklarının içinde diğer Markdown öğelerini kullanabiliriz.
> - Madde 1
> - **Önemli** Madde
