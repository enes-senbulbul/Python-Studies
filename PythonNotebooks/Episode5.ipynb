# İleri Fonksiyonlar ve Fonksiyonel Programlama

### Variable Length Arguments (*args)
- Fonksiyon tanımında parametrenin başına konulan yıldız * operatörü gönderlen positional(3 şekilde argüman gönderebiiyorduk, onlardan biri) argümanların hepsini toplayıp bir tuple içinde paketler (packing).
- args ismi bir gelenektir/convention PEP-8. Sihri yapan * operatörüdür.
- Matematikteki toplama operatörü $\sum$ gibidir. $f(x_1, x_2, ...,x_n)$ şeklinde n tabanlı bir vektörü kabul eder.

```python
# stardart fonk. Sadece 2 argümanı toplayabilir.
def topla_klasik(a, b):
    return a + b

# *args yöntemi (python arka planda gelen sayıları tuplu'a dönüştürür.)
def topla_vector(*args):
    print(f"Tip: {type(args)}") # args'ın tuple olduğunun kanıtı
    print(f"Paketlenen Veri: {args}")
    toplam = 0
    for sayi in args:
        toplam += sayi
    return toplam

# Testler 
print("Toplama klasik", sonuc1 := topla_klasik(10, 20))
print("-"*20)
print("Toplama *args ile: ", sonuc2 := topla_vector(10, 20, 30, 40, 50))


# Farklı veri tipleri ile args kullanımı (String birleştirme)
def birlestir(*args, ayirac =" "):
    return ayirac.join(args)

print("-"*20)
mesaj = birlestir("Python", "Matematik", "Mühendisliği", ayirac=" | ")
print(mesaj)
```
