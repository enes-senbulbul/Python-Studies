# Görev 1 ve 2: Kişisel Bilgi Kartı ve VKİ Hesaplayıcı

```python
ad = input("Adınızı girin:\n")
kilo = float(input("Kilonuzu girin (kg):\n"))
boy = float(input("Boyunuzu girin (m)[örn: 1.78]:\n"))

vki = kilo / boy**2
print(f"Sayın {ad}, hesaplanan VKİ değeriniz: {vki:.2f}")
print(ad, boy, kilo, sep=" | ")

if vki < 18.5:
    durum = "Zayıf"
elif vki < 25:
    durum = "Normal"
elif vki < 30:
    durum = "Kilolu"
else:
    durum = "Obez"

print(f"VKİ Değeriniz {vki:.2f} - Durumunuz: {durum}")
```

# Görev 3 ve 4: Listeler ve Döngü Kontrolü

```python
agirliklar = list()
while True:
    agirlik = input("Ağırlık girim (Çıkış için 'q'):")
    if agirlik.lower() == "q":
        print("Çıkış yapılıyor.")
        break
    agirlik = float(agirlik) 
    if agirlik <= 0:
        print("Hatalı veri, kaydedilmedi.")
        continue
    else:
        agirliklar.append(agirlik)

if agirliklar:
    print("Ağırlıklar:", agirliklar)
    print("Set sayısı:", len(agirliklar))
    max_deger = agirliklar[0]
    toplam = 0
    for a in agirliklar:
        toplam += a
        if a > max_deger:
            max_deger = a
    print("Kaldırılan en yüksek ağırlık:", max_deger)
    print("Kaldırılan ortalama ağırlık:", toplam/len(agirliklar))
    print("İlk üç ağırlık:", agirliklar[:3])
    agirliklar.sort()
    print("Küçükten büyüğe sıralanmış:", agirliklar)
```

# Görev 7: Fonksiyonlar ve Parametreler

```python
def performans_analizi(agirlik_listesi, limit=50):
    limit_exceed = [agirlik for agirlik in agirlik_listesi if agirlik >= limit]
    if len(limit_exceed): # ZeroDivisionError için Önlem
        average = sum(limit_exceed)/len(limit_exceed)
    return len(limit_exceed), average

sayi, ortalama = performans_analizi(agirliklar)
print(f"Limitten ağır ağırlık sayısı: {sayi}, ortalaması: {ortalama}")
sayi, ortalama = performans_analizi(agirliklar, 70)
print(f"Limitten ağır ağırlık sayısı: {sayi}, ortalaması: {ortalama}")
```

# Görev 9: Değişkenlerin Yaşam Alanı (Global & Local Scope)

```python
toplam_ziyaretci = 0
def turnike_gecis():
    global toplam_ziyaretci
    toplam_ziyaretci += 1
    anlik_voltaj = 12
    print(f"Turnike döndü. Toplam içerideki kişi: {toplam_ziyaretci}")

for i in range(3): turnike_gecis()
try:
    print(anlik_voltaj)
except NameError as hata:
    print(hata)
```

# Görev 11: Akıllı Sıcaklık Dönüştürücü

```python
while True: 
    sicaklik = input("Sıcaklık değeri girin:\n")
    try:
        sicaklik = float(sicaklik)
    except ValueError:
        print("Lütfen sayı girin...")
        continue
    while True:
        birim = input("Sıcaklığın birimini girin:\n")
        if birim.lower() not in ["f","c"]:
            print("Lütfen belirtilen birimlerden birisini girin...")
            continue
        else:
            break
    break

if birim.lower() == "f":
    sicaklik_f = sicaklik
    sicaklik_c = (sicaklik - 32)*(5/9)
else:
    sicaklik_c = sicaklik
    sicaklik_f = (sicaklik * 9/5) + 32

print(f"{sicaklik_c:.1f}°C = {sicaklik_f:.1f}°F")
```

# Görev 12: Öğrenci Not Sistemi (Listeler, Döngüler)

```python
notlar = list()
max_not = 0
min_not = 100
toplam = 0
notlar_50 = list()

for i in range(5):
    while True:
        ogr_not = input(f"Öğrenci {i+1}'in notunu girin (0-100):\n")
        try:
            ogr_not = float(ogr_not)
        except ValueError:
            print("Sayı girmen lazım. Tekrar gir.")
            continue
        if not (0 <= ogr_not <=100):
            print("0 ile 100 arasında bir değer girmen lazım. Tekrar gir.")
            continue
        else:
            notlar.append(ogr_not)
            toplam += ogr_not
            if ogr_not > max_not:
                max_not = ogr_not
            if ogr_not < min_not:
                min_not = ogr_not
            if ogr_not > 50:
                notlar_50.append(ogr_not)
        break

notlar.sort()
print("Sıralı notlar:", notlar)
print("Geçenler:", notlar_50)
print("En yüksek not:", max_not)
print("En düşük not:", min_not)
print("Sınıf ortalaması", toplam/5)
```

# Görev 13: DNA Dizisi Analiz Sistemi (String + Dictionary) 

```python
while True:
    seq = input("Bir DNA dizisi yaz")
    list_seq = list(seq)
    if list_unvalid:=[letter for letter in list_seq if letter.upper() not in ["A","T","G","C"]]:
        print("Geçersiz karakter var. Lütfen diziyi tekrardan girin.")
        print(f"Geçersiz karakterler: {list_unvalid}")
        continue
    else:
        seq_upper = seq.upper()
        print(f"DNA dizisi başarıyla girilmiştir. Dizi: {seq_upper}")
    break

list_seq = list(seq_upper)
num_A = len([letter for letter in list_seq if letter == "A"])
num_T = len([letter for letter in list_seq if letter == "T"])
num_G = len([letter for letter in list_seq if letter == "G"])
num_C = len([letter for letter in list_seq if letter == "C"])

seq_opposite = []
for letter in seq_upper:
    if letter == "A":
        seq_opposite.append("T")
    elif letter == "T":
        seq_opposite.append("A")
    elif letter == "G":
        seq_opposite.append("C")
    else:
        seq_opposite.append("G")
        
seq_opposite_str = "".join(seq_opposite)

print("DNA ANALİZ SONUÇLARI")
print("-"*30)
print(f"Dizi: {seq_upper} ({len(seq_upper)} baz)")

print("\nBaz Dağılımı:")
print(f"A: {num_A} (%{num_A/len(seq_upper)*100})")
print(f"G: {num_G} (%{num_G/len(seq_upper)*100})")
print(f"C: {num_C} (%{num_C/len(seq_upper)*100})")
print(f"T: {num_T} (%{num_T/len(seq_upper)*100})")

print(f"\nGC İçeriği: %{(num_G + num_C)/len(seq_upper)*100}")
print("Tamamlayıcı Zincir:")
print(f"Orijinal:\t {seq_upper}")
print(f"Tamamlayıcı:\t {seq_opposite_str}")
```
