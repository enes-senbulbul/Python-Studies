"#" işaretinden sonra mutlaka bir boşluk bırakmalıyız.

Normal metinler için özel bir işaret kullanmamıza gerek yoktur. Sadece yazmaya başlamalıyız.
Sonraki paragrafa geçmesi için mutlaka 1 satır boşluk bırakmalıyız. 
Yani 2 defa enter'a basmalıyız.

Aynı bu şekilde...
Eğer boşluk bırakmazsak metinler yan yana yapışır.
Github dahil çoğu markdown editöründe `enter` tuşuna bir defa basmak paragrafı bölmez. İki paragrafı ayırmak için iki kez enter tuşuna basmalıyız. Yani arada boş bir satır gerekir.
