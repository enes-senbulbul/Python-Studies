# Operatörler
- Öncelikle bi operatörleri detaylı çalışayım.
- Python'daki operatör çeşitleri:
    1. Aritmetik Operatörler (Arithmetic)
    2. Atama Operatörleri (Assignment)
    3. Karşılaştırma Operatörleri (Comparison)
    4. Mantıksal Operatörler (Logical)
    5. Kimlik Operatörleri (Identity)
    6. Üyelik Operatörleri (Membership)
    7. Bitwise Operatörler

### Aritmetik Operatörler
Bildiğimiz toplama, çıkarma, çarpma, bölme ve mod alma'nın yanında şöyle ekstralar var:
- üs alma için ** kullanılıyor.
- Floor division // operatörü (normal bölme float döndürürken bu integer döndürüyor.)

### Atama Operatörleri
- Bildiğimiz atama operatörü: =
- Aritmetik kombine: +=, -=, *=, /=, //=, %=, **=
- Bitwise kombine: &=, |=, ^=, >>=, <<=
- Walrus operatörü: ":=" Python 3.8 ile gelmiş. Daha kapsamlı expressionların içinde anlık atama yapmamızı sağlıyor.

```python
print(x:=3)

# Bununla aynı şey
x = 3
print(x)

```

### Karşılaştırma Operatörleri
- Yine bildiğimiz gibi: ==, !=,  >, <, >=, <=
- Ayrıca diğer programlama dillerinden farklı olarak zincirleyebiliyoruz.

```python
x = 5
print(1 < x < 10)

# Şununla aynı
print(1 < x and x < 10)

```

### Mantıksal Operatörler
- Bu sefer & ve | işaretleri yok. Bunlar bitwise operatör olarak ayrılmış.
- Pythonda direkt keyword olarak ayrılmışlar: "and", "or" ve "not" 

````python
x = 5
print(x>0 and x<10)
print(x<5 or x>10)
print(not(x>3 and x<10))
````

### Kimlik Operatörleri (Identity Operators)
- Nesneleri karşılaştırmak için kullanılır. İki tanedir: "is", "is not"
- "is" bellekte aynı lokasyonda bulunan aynı nesne'yi gösteren değişkenler için "True" değerini döndürür.
- Kullanım itibariyle == ile karıştırılmamalı. "is" direkt hafızada aynı lokasyonda bulunup bulunmadıklarına bakıyor. "==" ise değişkenlerin taşıdıkları değer(value)'ların eşit olup olmadıklarını kontrol ediyor.

````python
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z)
print(x is y)
print(x == y)
print(x is not y)
````

### Üyelik Operatörleri (Membership Operators)
- Bir nesne bir dizi'nin içinde var mı? kontrol etmek için kullanılır.
- İki tanedir: "not", "not in"
- Yine bool değer döndürür.
- Bu dizi -> string, list, tuple, range, set, frozenset, dict (sadece key'leri kontrol ediyor)

````python
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)
print(["apple", "cherry"] in fruits)
print({"apple", "cherry"} in fruits)
print(("apple", "cherry") in fruits)
print(["apple", "banana"] in fruits)
print({"apple", "banana"} in fruits)
print(("apple", "banana") in fruits, end="\n\n")

text = "Hello World"
print("H" in text)
print("He" in text)
print("Hello" in text)
print("Helo" not in text, end="\n\n")

print(1 in range(6))
fruits_set = {"apple", "banana", "cherry"} 
print({"apple", "cherry"} in fruits_set)
print({"apple"} in fruits_set) # Öyylesine denemeleeer
````
