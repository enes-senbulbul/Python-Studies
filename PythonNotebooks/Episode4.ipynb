# BÖLÜM 4: Fonksiyonlar 

### Fonksiyon Tanımlama (def)
- Fonksiyonlar kodları isimlendirilmiş, paketlenmiş ve tekrar kullanılabilir makinelere dönüştürür.
- Bir fonksiyonun üç parçası vardır.
    - Input (Arguments)
    - Body (Process)
    - Output (Return)
- Neden kullanırız?
    1. Don't Repeat Yourself Prensibi -> Aynı kodu tekrar tekrar yazma
    2. Modülarite -> Karmaşık bir problemi küçük parçalara ayır ve öyle çöz.

syntax:
```python
def fonksiyon_adi():
    # Yapılacak İşlemler 
    pass
```
print gibi fonksiyonlar -> Sadece ekrana yazı yazar, bir veri döndürmez.
return -> Fonksiyonun sonunda döndürmesini istediğimiz verileri tutan değişkenler


* CORE: Varsayılan değeri olan parametreler (Default Params) fonksiyon tanımlanırken her zaman en sonra yazılmalıdır. def test(a=1, b): (YANLIŞ: SyntaxError), def test(b, a=1): (DOĞRU)

````python
# Void Fonksiyon -> Değer geri dndürmeyen, sadece iş yapan fonksiyonlardır.
def sistem_bilgisi_yazdir():
    print("--- SİSTEM BAŞLATILIYOR")
    print("Versiyon 1.0.4")

sistem_bilgisi_yazdir() # Fonksiyon Çağırma

# Değer Döndüren Fonksiyon (return Kullanan)
def kinetik_enerji_hesaplama():
    # Şimdilik parametre almasın
    kutu_bilesen = 10 # kg
    hiz_vektoru = 5 # m/s
    enerji = 0.5 * kutu_bilesen * (hiz_vektoru**2)
    return enerji # Sonucu çağıran yere fırlat

sonuc = kinetik_enerji_hesaplama()
print(f"Hesaplanan Enerji: {sonuc} Joule")
````

### Arguments & Parameters
- Parameter: Fonksiyonu tanımlarken parantez içine yazdığımız değişken isimleri (x, y gibi)
- Argument: Fonksiyonu çağırırken gönderdiğimiz gerçek değerler (5, "Ali" gibi)

Python'da 3 temel argüman gönderim şekli vardır.
1. Positional (Sıralı): Sıraya göre eşlenir. İlk giden ilk değişkene oturur. 
2. Keyword (İsimli): Sıra önemsizdir, ismiyle çağrılır (yas=21 gibi).
3. Default (Varsayılan): Değer gönderilmezse standart bir değer kullanır.

````python
# Parametrelerden birisi zorunlu diğeri ise opsiyonel
def fatura_hesapla(aktif_kullanim, birim_fiyat = 1.5):
    tutar = aktif_kullanim * birim_fiyat
    print(f"Tüketim: {aktif_kullanim} kWh, Tarife: {birim_fiyat} TL -> Tutar: {tutar} TL")

fatura_hesapla(100, 2.0) #Sıralı Argümanlar
fatura_hesapla(birim_fiyat = 3.5, aktif_kullanim = 50) # Keyword Argümanlar
fatura_hesapla(200) # Default Argüman

# Hata Analizi
try:
    fatura_hesapla()
except TypeError as hata:
    print("aktif_kullanim'a varsayılan değer atanmamış, mecbur girmelisin!")
    print("HATA MESAJI:",hata)
````

### Return Values & Scope 
Scope(Kapsam): Her değişkenin bir yaşam alanı vardır.
- Local Scope: Fonksiyonun içinde tanımlanan değişkenlerdir.
- Global Scope: En tepede tanımlanan herkesin görebildiği değişkenlerdir.
* CORE: Bir fonksiyonun içinde tanımlanan değeri return etmezsek, o değer fonksiyon bitince yok olur.
* PITFALL: Fonksiyon içinde global değişkene yeni bir değer atamaya kalkışırsak `örn: global_degisken = 5`, Python global değişkeni değiştirmez sadece aynı isimde başka bir nesneyi gösteren yeni bir yerel değişken yaratır. Global'i gerçekten değiştirmek için `global` keyword'ü gerekir ama önerilmez. 

````python
global_sicaklik = 25

def isi_hesapla(kutle, oz_isi, delta_t):
    # Kel Macit Formülü
    enerji = kutle * oz_isi * delta_t
    # enerji burada local değişkendir. Sadece bu blokta yaşar.
    yeni_sicaklik = global_sicaklik + delta_t
    # Global değişkene erişim
    print(f"[Log Kaydı] İçerideki hesap: {enerji} Joule, Son Sıcaklık: {yeni_sicaklik}")
    
    # ÇOKLU Return -> Tuple olarak döner(packing) 
    return enerji, yeni_sicaklik

Q, T_son = isi_hesapla(kutle=2, oz_isi=4.18, delta_t=10) # Sonuçları yakalıyoruz (Unpacking)

print("\n--- Sonuçlar ---")
print(f"Transfer Edilen Isı: {Q}")
print(f"Sistemin Son Sıcaklığı: {T_son}")

# Scope Testi (Hata Bekliyoruz)
try: 
    print(enerji)
except NameError as hata:
    print("eneji fonksiyonun içinde doğdu ve fonksiyon bitince öldü. Local değişkendi.")
    print("HATA MESAJI:", hata)
````

# Challange Me - Akıllı Fabrika Kalite Kontrol Sistemi

````python
import random

def veri_olustur(adet=10):
    borular = []
    for i in range(adet):
        borular.append(dict(id_num = 101 + i, uzunluk = random.uniform(90,110), hata_orani = random.uniform(0, 0.1)))
    return borular
    
def kalite_kontrol(sozluk, min_uzunluk=100, max_hata=0.05):
    if (sozluk["uzunluk"] >= min_uzunluk) and (sozluk["hata_orani"] < max_hata):
        onay = True
    else:
        onay = False
    return onay
    
def ayiklama_yap(urun_listesi):
    kabul_edilenler = list()
    hurda = list()
    for urun in urun_listesi:
        durum = kalite_kontrol(urun)
        if durum == True:
            kabul_edilenler.append(urun)
        else:
            hurda.append(urun)
    return kabul_edilenler, hurda

liste = veri_olustur()
kabuledilen, hurdacikan = ayiklama_yap(liste)
print(f"Toplam Üretim: {len(liste)} adet. Kabul Edilen: {len(kabuledilen)} adet. Başarı Oranı: %{(len(kabuledilen)/len(liste))*100}")
````
