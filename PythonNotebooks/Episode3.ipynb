### Lists (Listeler) - Oluşturma ve Erişim
- Şu ana kadar değişkenlerde tek bir değer tuttuk. Örn: x = 5
- Ancak gerçek dünyada veriler gruplar halinde gelir. Örn: bir sınıftaki notlar, sensörden gelen 100 ölçüm
- Listeler SIRALI (ORDERED) ve DEĞİŞTİRİLEBİLİR (MUTABLE) veri koleksiyonlarıdır.
- Python'da listeler heterojendir. Yani aynı liste içinde hem sayı hem metin hem de başka bir liste daha barındırabilir. Yani bu özelliğiyle C veya Java'daki Array'lerden çok daha esnektir. Ayrıca listelerin boyutları da dinamiktir. Preallocation yapmamıza da gerek yok.
- IndexError -> Olmayan bir sıraya erişmeye çalışırsak -3 elemanlı listede 10. elemana mesela- program çöker.

````python
# Liste Oluşturma
collatz_sequence = [10, 5, 16, 8, 4, 2, 1] # Homojen Liste, sadece int var
student_info = ["Ali Yılmaz", 2023001, 3.85, True] # Heterojen Liste
print("Dizi: ", collatz_sequence)
print("Öğrenci: ", student_info)

# Erişim (Indexing) - Stringlerdekiyle aynı mantık
first_term = collatz_sequence[0]
last_term = collatz_sequence[-1]
print(f"\nBaşlangıç: {first_term}, Bitiş: {last_term}")

# İç içe Listeler (Nested Lists) -> Matris Gösterimi
matrix = [[1, 2], [3, 4]]
element_21 = matrix[1][0] # 2.satır 1. sütundaki eleman
print("Matris Elemanı (2, 1):", element_21)

# Hata Testi
try:
    print(collatz_sequence[100])
except IndexError as hata:
    print("IndexError hatası verdi.")
    print("HATA MESAJI:", hata)
````

### Lists (Listeler) - Methodlar ve Değiştirilebilirlik
- Stringlerde Indexing özelliği olsa da Listeler ayrıca stringlere nazaran "DEĞİŞTİRİLEBİLİR" yapıdadır. Yani hafızadaki yerini değiştirmeden içeriğini -elemanlarını- değiştirebilir, silebilir veya yeni eleman ekleyebiliriz.
- Kritik methodlar:
    * .append(x): Listenin en sonuna x elemanını ekler. (En sık kullanılan)
    * .insert(i, x): Belirtilen i endeksine x elemanını sıkıştırır, sonrakileri sağa kaydırır.
    * .pop(i): Belirtilen indeksteki elemanı listeden siler ve o elamanı döndürür. İndeks verilmezse sonuncuyı atar ve bize verir.
    * .remove(x): Değeri x olan ilk elemanı bulur ve siler.
- Liste methodlarının çoğu işlemi in-place yapar. Yani işlemi liste içine yapar ve geriye None döndürür. Ve yani yeni_liste = eski_liste.sort() dersek yeni_liste ddeğikenimiz None -boş- olur.
- PITFALL: The Aliasing Trap (Alias: takma ad): list_a'yı oluşturup list_b = list_a dersek list_b'yi değiştirdiğimizde list_a da değişir. Çünkü ikisi de bellekte aynı listeyi gösteren etiketlerdir. list_b, list_a'nın kopyası değil onun takma adıdır(alias).

````python
grades = [70, 85, 90, 45]
print("İlk Hali:", grades)

# Indeks ike doğrudan değiştirme 
grades[3] = 50
print("Güncel:", grades)

# Ekleme (Append ve Insert)
grades.append(100) # Sona ekleme
grades.insert(0, 95) # Başa ekleme
print("Eklemeler Sonrası:", grades)

# Silme (Pop ve Remove)
silinen = grades.pop() #Son elemanı siler ve döndürür.
print(f"Silinen Not: {silinen}")
print("Pop Sonrası:", grades)
grades.remove(70)
print("Remove Sonrası:",grades)

# Kopyalama Tuzağı (Aliasing Pitfall)
list_a = [1, 2, 3]
list_b = list_a
list_b[0] = 999
print(f"List B: {list_b}")
print(f"List A: {list_a}")

# Güvenli Kopya Nasıl Alınır?
list_c = list_a.copy()
list_d = list_a[:]
list_c[0] = 1
print(f"List C: {list_c}")
print(f"List A: {list_a}")
````

### Tuples (Demetler/Çoklular)
- Listeler köşeli parantez ile oluşturulur ve değiştirilebilirler.
- Tuple(Demet) ise yuvarlak parantez ile oluşturulur ve DEĞİŞTİRİLEMEZ.
- Oluşturduktan sonra ne eleman ekleyebilir ne de silebiliriz.
- Bir **MatMüh** için Tuple:
    * Vektör sabitleri: Değiştirilmemesi gereken koordinatlar (x, y, z)
    * Veri Kayıtları: Bir veritabanı satırı (ID, İsim, Tarih) gibi yapısı bozulmaması gereken gruplar.
- Neden Tuple Kullanırız?
    * Veri Güvenliği: Tanlışlıkla verinin silinmesini engeller.
    * Performans: Listelerden daha hafiftir ve daha hızlı erişilir.
    * Hashable: Dictionary anahtarı olarak kullanılabilirler. !!! Listeler kullanılamaz.

```python
koordinat = (3, 4)
renk = (255, 0, 0) # RGB kırmızı
```

- CORE: Tuple'ların en güçlü yanı değerleri paketleyip (packing) daha sonra tek hamlede değişkenlere dağıtabilmesidir (unpacking).
- PITFALL: Tek elemanlı bir tuple oluştururken sonuna virgül koymamız lazım. Bunu yapmazsak eğer python bunu parantez kullandığımız normal bir işlem sanar ve tuple değil de int/float vs. olarak atar.

```python
x, y = (3, 4) # Burada x=3, y=4 olur.
a = (5) # a integerdır.
b = (5, ) # b ise tuple
```

````python
# Tuple Oluşturma 
point_a = (3, 4, 5)
print(f"Nokta A: {point_a}, Tip: {type(point_a)}")

# Immutability Test (Değiştirilemezlik Testi)
try:
    point_a[0] = 10
except TypeError as hata:
    print("Tuple Immutable'dır. İçeriğini değiştiremeyiz.")
    print("HATA MESAJI:", hata)

# Tuple Unpacking (Mühendislikte çok yaygın)
x, y, z = point_a # Vektör bileşenleri ayırma
print(f"x: {x}, y: {y}, z: {z}")

# Geçici Değişken Kullanmadan iki değişkeni takas etme  
# Swap İşlemi (Pythonic Way)
a = 10
b = 20
a, b = b, a # Arka planda tuple unpacking kullanılır (20, 10)
print(f"a: {a}, b: {b}")

# Tek Elemanlı Tuple Tuzağı
not_tuple = (50)
real_tuple = (50,)
print(f"Yanlış: {type(not_tuple)}, Doğru: {type(real_tuple)}")
````

### Dictionaries (Sözlükler)
- Listelerde verilere ulaşmak için indexler(0, 1, 2) kullanıyorduk.
- Sözlük yapısında ise verilere bizim belirlediğimiz **Anahtarlar(Keys)** ile ulaşırız.
- **Teknik Derinlik:** Arka planda *Hash Table* yapısı kullanır. Bu sayede listenin boyutu ne olursa olsun (ister 10 eleman, ister 1 milyon eleman), bir veriye erişim süresi teorik olarak sabittir.
- Listelerde arama yapmak O(n) iken sözlüklerde anlıktır O(1).
- Sözlük oluştururken süslü parantez ve iki nokta kullanırız.

```python
sozluk = {
    "anahtar1": "deger1",
    "anahtar2": 100,
    "anahtar3": [1, 2, 3] #Değer her şey olabilir.
}
```

- CORE: Sözlükteki anahtarlar(keys) mutlaka Immutable ve Unique olmalıdır. (Değişmez ve eşsiz.)
- String, numeric types, tuple anahtar olabilir ama list OLAMAZ.
- Değerler(values) için ise HER ŞEY olabilir. Liste de...
- PITFALL: Olmayan bir anahtara doğrudan erişmeye çalışırsak (örn: sozluk["olmayan"]) program çöker (KeyError). Bunun yerine .get() metodunu kullanmak hayat kurtarır. 

````python
# Sözlük Oluşturma
student = {
    "ad": "Enes",
    "soyad": "Şenbülbül",
    "no": 90240735,
    "notlar": [80, 75, 50],
    "aktif": True
}
print("Öğrenci Profili:", student)

# Veriye Erişim (Lookup)
print(f"Adı: {student["ad"]}")
try:
    print(f"Bölüm: {student["bolum"]}")
except KeyError as hata:
    print("Olmayan anahtara doğrudan erişmeye çalıştığımız için KeyError verdi.")
    print("HATA MESAJI:",hata)

# Güvenli Erişim (.get() Methodu)
bolum = student.get("bolum", "Bölüm Girilmemiş")
# Varsa getirir, yoksa None veya ayarlanan değeri döndürür ve HATA VERMEZ.
print(f"Bölümü: {bolum}")

# Veri Ekleme ve Güncelleme
student["bolum"] = "Matematik Mühendisliği" # Yeni anahtar-değer ekleme
student["no"] = 9024073502 # var olan anahtara karşılık gelen değeri güncelleme
print("Güncel Profil:", student)

# Sözlük Üzerinde Gezinme (Iteration)
print("\n--- Kimlik Kartı ---")
for key, value in student.items():
    print(f"{key.capitalize()}: {value}")
````

### Sets (Kümeler)
- Tekrarlayan verileri otomatik olarak temizleyen özel bir yapıdır.
- Eşsiz(UNIQUE), sırasız(UNORDERED) elemanların bir koleksiyonudur.
- Kritik Özellikler:
    - Uniqueness: Bir küme içerisinde aynı elemandan iki tane olamaz (Tekrarları otomatik olarak siler).
    - Unordered: Indeks kavramı yoktur. kume[0] diyemeyiz.
    - Hızlı erişim: Sözlükler gibi *Hash Table* kullanır. Kümelerde arama yapmak da anlıktır O(1).
- Küme operasyonları:
    - Birleşim (A u B): A | B veya A.union(B)
    - Kesişim (A n B): A & B veya A.intersection(B)
    - Fark (A \ B): A - B veya A.difference(B)
    - Simetrik Fark (A $\Delta$ B):  A ^ B (Her ikisinde de olmayan ama birleşimde olanlar)
- CORE: Boş bir sözlük oluşturmak için {} süslü parantezleri tek başına kullanamayız. Bu boş sözlük oluşturur. Mutlaka set() fonksiyonunu kullanmalıyız.
- PITFALL: Set elemanları da (keys) gibi immutable olmalıdır. Bir kümenin içine LİSTE KOYAMAYIZ. 

````python
# Kümeleri Oluşturma
calculus_students = {"Ali", "Ayşe", "Mehmet", "Deniz"}
algebra_students = {"Mehmet", "Can", "Deniz", "Elif"}
print(f"Calculus: {calculus_students}")
print(f"Algebra: {algebra_students}")

# Kesişim, Birleşim, Fark
both_courses = calculus_students.intersection(algebra_students)
all_students = calculus_students.union(algebra_students)
only_calc = calculus_students.difference(algebra_students)
one_course = calculus_students.symmetric_difference(algebra_students)
print(f"İki Dersi de Aynı Anda Alanlar: {both_courses}")
print(f"Aynı anda iki ders almayanlar: {one_course}")
print(f"Tüm öğrenciler: {all_students}")
print(f"Sadece Calculus alanlar: {only_calc}")

# Mühendislik Hilesi: Deduplication (Tekrarları Silme)
raw_sensor_data = [10, 20, 10, 30, 20, 40, 10]
unique_data = list(set(raw_sensor_data))
print(f"Ham Veri: {raw_sensor_data}")
print(f"Temizlenmiş Veri: {unique_data}")
````

### Gemini Çözüm

````python
import random
from datetime import date

fiyat_havuzu = [10000, 20000, 3000, 8000, 15000, 7000]
stok_havuzu = [5, 2, 4, 3, 6, 4]

tedarikci_A = ["Elektronik", "Bilgisayar", "Laptop", "Elektronik"]
tedarikci_B = ["Laptop", "Ofis", "Kırtasiye", "Bilgisayar"]

set_A = set(tedarikci_A)
set_B = set(tedarikci_B)
tum_etiketler = set_A | set_B
print(f"Birleştirilmiş Kategori Listesi: {tum_etiketler}")

# GÖREV 1: Envanter Oluşturma
envanter = {}
k = 1
for kategori in tum_etiketler:
    urun_id = f"PRD-{k:03d}" # f_string ile ID formatlama
    envanter[urun_id] = {
        "ad": kategori + " Modeli", # Süsleyelim demiş Gemini
        "fiyat": random.choice(fiyat_havuzu),
        "stok": random.choice(stok_havuzu)
    }
    k += 1
    
print("\n--- Oluşturulan Envanter ---")
for id_numarası in envanter:
    print(id_numarası, envanter[id_numarası])

# GÖREV 2: Satış Simülasyonu (Tuple)
satis_gecmisi = []
for islem_no in range(1, 10):
    mevcut_id_listesi = list(envanter.keys())
    satilan_id = random.choice(mevcut_id_listesi)
    bugun = date.today()
    kayit_tuple = (islem_no, bugun, satilan_id)
    satis_gecmisi.append(kayit_tuple)
    
print("\n--- Satış Geçmişi (Tuple Listesi) ---")
for satis in range(0, len(satis_gecmisi)):
    print(satis_gecmisi[satis])

# GÖREV 3: Pahalı Ürünlerin Analizi (List Comprehension)
pahali_urunler = [
    detay["ad"]
    for urun_id, detay in envanter.items()
    if detay["fiyat"] > 10000
]

print("\n--- Lüks Ürünler (>10.000 TL) ---")
print(pahali_urunler)
````
