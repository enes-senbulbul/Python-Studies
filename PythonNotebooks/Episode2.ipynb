# Kontrol Yapıları (Control Flow)
- Şu ana kadar yazdığımız kodlar yukarıdan aşağıya şelale gibi aktı (Linear Flow).
- if blokları ise programa karar verme yetisi kazandırır.
- C++ ve Java'da süslü parantezler kullanılırken Python'da bloklar girinti (indentation) ile belirlenir.
- Python'da boşluklar (whitespace) syntax'ın bir parçasıdır. Yanlış girinti IndentationError verir. Standart 4 boşluktur.

```python
course_grade = 79.0
attendance = 85
print(f"Öğrenci Notu: {course_grade}")

# Senaryo: Harf Notu Hesaplama
if attendance < 70:
    letter_grade = "NA"
    status = "Başarısız (Devamsız)"
else:
    if course_grade >= 90:
        letter_grade = "AA"
        status = "Mükemmel"
    elif course_grade >= 85:
        letter_grade = "BA"
        status = "Çok İyi"
    elif course_grade >= 80:
        letter_grade = "BB"
        status = "İyi"
    elif course_grade >=70:
        letter_grade = "CB"
        status = "Orta"
    elif course_grade >=50:
        letter_grade = "DD"
        status = "Koşullu Geçer"
    else:
        letter_grade = "FF"
        status = "Kaldı"

print(f"Harf Notu: {letter_grade}")
print(f"Durum: {status}")
```

### Truthy and Falsy Values
- Python'da koşullu ifadelerin içine sadece True veya False yazmak zorunda değiliz. Python'da her nesnenin bir doğruluk değeri vardır.
- Falsy(Yanlışsı)
    1. None
    2. False
    3. Sayısal Sıfırlar: 0, 0.0, 0j
    4. Boş diziler: ""boş string, []boş liste, ()boş tuple
    5. Boş sözlük: {}

> Python Essential: Bir listenin boş olup olmadığını kontrol etmek için `if len(liste)>0:` diye kontrol etmektense `if liste:` diye kontrol etmek daha profesyoneldir.

>  Pitfall: Eğer değişken kod çalışırken 0 değerini alabiliyorsa (örn: sıcaklık $^\circ$C) `if sicaklik:` derseniz Python bunu False sayar ve kod çalışmaz. Onun yerine `if sicaklik is not None:` demek gerekir.

```python
# Boş string kontrolü (Hata Mesajı)
error_message = "" # Şu an hata yok. Boş = Falsy -> Bu sebeple 0 olup else bloğunu çalıştırır.

if error_message:
    print(f"SİSTEM HATASI: {error_message}")
else:
    print("Sistem stabil çalışıyor. (Hata mesajı boş)")

# Liste Kontrolü (Veri Akışı)
sensor_buffer = [24.5, 24.8, 25.1] # Dolu Liste = Truthy
if sensor_buffer:
    print(f"İşlenecek {len(sensor_buffer)} adet veri var.")
else:
    print("Buffer boş, veri bekleniyor...")

# Sayısal 0 tuzağı
battery_level = 0 # Pil tamamen bitik ama veri var.

if battery_level: # Yanlış Kullanım
    print("Pil durumu okundu.")
else:
    print("Pil verisi alınamadı!")

if battery_level is not None: # Doğru Kullanım
    print(f"Pil durumu: %{battery_level}")
```

# Döngüler (Loops)
- Programın aynı işi birden fazla kez tekrar etmesini sağlar.

### while Loops
- if yapısının sürekli tekrar eden halidir. Belirtilen koşul True olduğu sürece bloğun içindeki kodları tekrar tekrar çalıştırır.
- Nerede duracağını belirtmezsek döngü sonsuza kadar gider.
- While genellikle adım sayısı önceden bilinmeyen durumlarda kullanılır.
    * Kullanıcı doğru şifreyi girene kadar sor.
    * Hata payı 0.001'in altına düşene kadar hesapla.
- Infinite Loop: Eğer döngü bloğu içinde koşulu False yapacak bir değişiklik yoksa program sonsuza kadar çalışır ve kilitlenir(Ctrl + C ile durdurmamız gerekir).

```python
# Hatırlatma: battery_level -> 0 idi.

print(f"Şarj işlemi başlıyor... (Mevcut: %{battery_level})")
while battery_level < 100:
    battery_level = battery_level + 10
    print(f"Şarj ediliyor... Seviye: %{battery_level}")
    if battery_level == 80:
        print("--- Hızlı şarj bitti, yavaş moda geçiliyor. ---")

print("Şarj Tamamlandı!")
```

### for Loops
- While'ın aksine veri analizi ve mühendislikte genellikle for kullanırız. For ile belirli bir veri seti üzerinde gezebiliriz.
- C, C++ ve Java'daki for(int i=0; i<10; i++) yapısının aksine Python'daki for döngüsü teknik olarak bir iterator(yineleyici)'dir. Bir koleksiyonun (liste, string veya dizi) elemanları üzerinde tek tek gezinir.
- range() fonksiyonu sayısal döngüler kurmak için aritmetik diziler üreten bir jeneratördür. String slicing mantığıyla aynı parametreleri alır. Sırasıyla start, stop, step. Stop dahil değildir ve step opsiyoneldir.
- CORE: range() fonksiyonu bellekte devasa bir liste oluşturmaz. Sadece ihtiyaç duyulduğunda bir sonraki sayıyı üreten bir "kural" (generator object) saklar. Bu sayede range(1000000000) bile belleği şişirmez.

```python
# Senaryo 1: Öğrenci ID oluşturma 
# 2023 girişli ilk 5 öğrenciye numara verelim.

print("--- Öğrenci ID Ataması---")
for i in range(1, 6):
    student_id = 20230000 + i
    print(f"Sıra {i}: Öğrenci ID {student_id}")


# Senaryo 2: Matematiksel Toplam (Sigma)
# 1'den 10'a kadar olan sayıların karelerini toplayalımç
sum_squares = 0
for i in range(1,11):
    square = i**2
    sum_squares += square
print(f"\nKareler Toplamı (1-10): {sum_squares}")

# Senaryo 3: String üzerinde gezinme (Iteration)
# Bir DNA dizisindeki "A" (Adenin) sayısını bulalım.
dna_sequence = "AGCTTAGCTA"
adenine_count = 0
for base in dna_sequence:
    if base == "A":
        adenine_count += 1
print(f"\nDNA Dizisi: {dna_sequence}")
print(f"Adenin Sayısı: {adenine_count}")
```

### Break & Continue 
- Normalde döngüler şart sağlanana (while) veya elemanlar bitene kadar (for) çalışır.
- Bazen bu akışı manuel olarak bozmamız gerekir.
    - `break`: Döngüyü tamamen öldürür. Örn: Aradığımız şeyi bulduk artık aramaya gerek yok.
    - `continue`: Sadece o anki turu pas geçer. Döngüyü bitirmez, başa döner ve bir sonraki elamandan devam eder. Örn: bu veri hatalı, bunu işleme ama diğerlerine bak.

```python
# Sensör Veri Akışı Simülasyonu 
# -1 değeri gelirse: Hatalı okuma, bunu atla (continue)
# 999 değeri gelirse: Acil durdurma sinyali, sistemi kapat (break)
# Diğer değerler: Ekrana bas

sensor_data_stream = [22, 23, -1, 24, -1, 25, 999, 26, 27]
print("--- Veri Akışı İşleniyor ---")
for data in sensor_data_stream:
    if data == 999:
        print("ACİL DURUM SİNYALİ: Sistem Kapatılıyor!")
        break
    if data < 0:
        print("Hatalı Veri Atlandı!")
        continue
    print(f"Okunan Değer: {data}")
print("--- İşlem Sonlandı ---")
```

# Challange me: Collatz Sanısı (3n+1 Problemi) Simülatörü
- Herhangi bir pozitif tam sayı al.
- Sayı çift ise 2'ye böl.
- Sayı tek ise 3 ile çarp ve 1 ekle.
- Bu işlemi sayı bir olana kadar tekrarla. Sanıya göre hangi sayıyla başlarsan başla sonuç mutlaka 1'e ulaşır.

```python
flag = 1
while flag:
    flag = 0
    sayi = int(float(input("Pozitif bir tam sayı gir.")))
    if sayi <= 0:
        print("Pozitif bir sayı girmen gerekiyor!")
        flag = 1

adim_sayisi = 0
max_deger = sayi
sayi_baslangic = sayi

while sayi != 1:
    if (sayi%2 == 0):
        print(sayi)
        sayi = sayi//2 # Floor division normal divisiondan farklı olarak int döndürür.
    else:
        print(sayi)
        sayi = 3*sayi + 1
    if max_deger < sayi:
        max_deger = sayi
    adim_sayisi += 1

print("--- İşlem Sona Erdi ---")
print(f"Başlanılan Sayı: {sayi_baslangic}\nToplam Adım Sayısı: {adim_sayisi}\nGörülen En Yüksek Değer: {max_deger}")
```

```python
# Optimize Edilmiş Kod (Gemini)

while True: # Input Validation - Pythonic Way: While True + break
    try:
        giris = input("Pozitif bir tam sayı giriniz: ")
        sayi = int(float(giris))
        if sayi > 0:
            break
        print("Lütfen pozitif bir sayı girin!")
    except ValueError:
        print("HATA: Sadece sayısal değer giriniz.")

sayi_baslangic = sayi
max_deger = sayi
adim_sayisi = 0

print(f"\n--- Yolculuk Başlıyor: {sayi} ---")
while sayi != 1:
    if sayi % 2 == 0:
        sayi = sayi//2
    else:
        sayi = 3*sayi + 1
    if sayi > max_deger:
        max_deger = sayi
    adim_sayisi += 1

print("-"*30)
print(f"Başlangıç: {sayi_baslangic}")
print(f"Toplam Adım: {adim_sayisi}")
print(f"Zirve Noktası (Peak): {max_deger}")
```
