### Indentation
- Python'da girintiler(indentations) kod bloklarını belirtmek için kullanılır. Diğer programlama dillerinde ise whitespace sadece okunabilirliği arttırmak için tercih edilir ve kod blokları süslü parantezlerle oluşturulur.
- Pythonda bir kod bloğunun içideki bütün satırlarda girintiyi eşit boşluk sayısıyla (genellikle 4) oluşturmamız gerekiyor. Aksi halde hata verir.
- Python'da iki statement'ı noktalı virgül ile ayırıp tek satırda yazabiliriz ama bu okunabilirliği azaltır.
- Python'da print() fonksiyonu varsayılan olarak yeni bir satırla sona erer. Ama bunu değiştirebiliriz. 

```python
# Indentation Hatası verir.
"""
if 5 > 2:
print("Five is greater than two!")
"""

# Indentation Hatası vermez.
if 5 > 2:
    print("Five is greater than two!")
elif 5 < 2:
  print("Five is greater than two!")

# Indentation Hatası verir. (Tek blok içinde farklı intendationlar)
'''  
if 5 > 2:
 print("Five is greater than two!")
        print("Five is greater than two!")
''' # Multiple line'ı tekli tırnaklarla da oluşturabiliriz.

# Semicolons
print("Hello"); print("How are you?"); print("Bye bye!")

# Same line
print("Hello", end=" ") # Sonraki satırın yapışmaması için tırnaklar arasına bir boşluk bırakmalıyız.
print("How are you?", end=" ")
print("Bye bye!", end=" ") 
```

### Integer & Floats (Numeric Types)
- Python'da üç tane numeric veri tipi var: int, float, complex
- Integers: Tam sayılar (Python 3'te int tipi sınırsız büyüklüktedir. Sonsuz büyüklükte sayı atayabiliriz.)
- Floats: Ondalıklı sayılar (Bilimsel gösterim için "e/E" harfi ile de oluşturulabilir.)
- Complex: Karmaşık sayılar (Sanal kısmı "j" harfi ile yazılır.)
> Python'da bölme işlemi (/) her zaman float döndürür. 10/2 -> 5.0

##### Aritmetik Operatörler
toplama<br>
çıkarma<br>
çarpma<br>
bölme -> Her zaman float döner. <br>
Modulus (%) -> 14 % 5 = 4 -> Bölümden kalanı verir.<br>
Floor Division (//) -> 14 // 5 = 2 -> Küsuratı atar.<br>
Üs Alma (**) -> `Diğer yazılım dillerindeki gibi ^ değil dikkat et. `

##### Atama Operatörü ile kombinasyonları
+=, -=, *=, /=, //=

```python
midterm = 70
final = 85
weight_midterm = 0.4
weight_final = 0.6
course_grade = (midterm * weight_midterm) + (final * weight_final)
print("Yıl Sonu Notu: ", course_grade)
print("Notun Tipi: ", type(course_grade))
print(isinstance(course_grade, float))

total_seconds = 135
minutes = total_seconds // 60
seconds = total_seconds % 60
print("Süre:", minutes, "dakika", seconds, "saniye")

huge_number = 2 ** 100
print("int veri tipi sonsuz büyüklükte sayı tutabilir. PYTHON'un GÜCÜ -> 2 üzeri 100: ", huge_number, end="\n\n")

# Floats as scientific numbers
x = 35e3
y = 12E-4
z = -87.7e100
print(x, y, z, sep="\t\t")
print(type(x), type(y), type(z), sep="\t")

# Complex numbers
number1 = 3j 
number2 = 2 - 1j #j'nin önüne 1 koymak zorundaymışız
print(str(number2))
print(number2)
```

### Random Sayı Oluşturma (Random Modülü)
- python'un kendisinde random sayı oluşturmak için bir fonksiyon yok. Ama built-in modülü olan random'dan fonksiyon çağırabiliyoruz.
- random.randrange(a, b, step): İki integer arasında random sayı üretir. a DAHİL, b DEĞİL, step opsiyonel
- random.randint(a,b): İki integer arasında random sayı üretir. a ve b DAHİL. Kapalı aralık. step yok.
- random.uniform(a,b): a ile b arasında float döndürür.
- random.random(): 0 ve 1 arasında değer döndürür. 0 DAHİL 1 DEĞİL.
- random.choice(seq): verilen itarable içidden rastgele eleman döndürür. list, tuple, string EVET. set, dict HAYIR çünkü sıralı değiller.
- random.choices(seq, k=n): Tekrarlı seçim. Liste döndürür.
- random.sample(population, k): Tekrarsız seçim. Kombinasyon mantığı. k adet seçer. Liste döndürür.
- random.shuffle(list): listeyi yerinde(in-place) karıştırır ve None döndürür.
- random.seed(42): Rastgeleliği tekrarlanabilir yapar.

### Strings - Temeller 
- Stringler karakterlerin sıralı bir dizisidir(sequence). Python'da metinler tek veya çift tırnak işaretleri arasına yazılır.
- Kritik Özellikler:
    - 0-based indexing: C'deki gibi 0'dan başlar.
    - Negative Indexing: Python'a özel bir yetenek. -1 son elemanı, -2 ise sondan bir önceki elemanı temsil eder.
    - Immutability (Değişmezlik): Bir string oluşturulduktan sonra içindeki tek bir harfi değiştiremeyiz. Değişiklik için yeni bir string oluşturmalıyız.
* Slicing (Dilimleme):<br>
string[start : stop : step]<br>
start -> başlangıç indeksi (dahildir.)<br>
stop -> bitiş indeksi (DAHİL DEĞİLDİR.)<br>
step -> atlama adımı (opsiyoneldir.)

* len() fonksiyonu boşlukları da birer karakter olarak sayar. 

```python
# Hatırlatma: course_name -> "Matematik Mühendisliği 101"

# Indexing (Erişim)
first_char = course_name[0]
last_char = course_name[-1]
print(f"İlk: {first_char}, Son: {last_char}")

# Slicing (Dilimleme)
first_word = course_name[0:9] # Matematik (0->M dahil) (8->t dahil)
jumpy = course_name[0:9:2]
print("Dilim:", first_word)
print("Atlamalı:", jumpy)

# Concatenation (Birleştirme)
greeting = "Merhaba" + first_word
print(greeting)

# Immutability 
try: 
    course_name[0] = "K"
except TypeError as message:
    print("Python'da string'in içinden harf değiştiremeyiz.")
    print("HATA:", message)

# len() Fonksiyonu - Uzunluk
length = len(course_name)
print("Karakter Uzunluğu:", length)
```

### Booleans & Logical Operators
- Python'da Boolean değerlerin baş harfi büyük yazılır: `True`, `False`
- Karşılaştırma Operatörleri: `==`, `!=`, `>`,`<`,`>=`,`<=`
- Bu operatörler çalışınca işlem sonucu olarak geriye her zaman bir bool (True/False) döndürürler.
- Mantıksal operatörler: `and`,`or`,`not`
- Sonucu boolean expressionları karşılaştırırlar ve yine bool (True/False) döndürürler.

```python
# Hatırlatma: course_grade -> 79.0

# Basit Karşılaştırma
is_passed = course_grade >= 50
print("Geçti mi?:", is_passed)

# Karmaşık Mantık
attendance = 85
has_paid_tuition = True
can_graduate = (course_grade >= 50) and (attendance >= 70) and (has_paid_tuition == True)
print("Mezuniyet Durumu:",can_graduate)

# "not" Operatörü
is_failed = not is_passed
print("Kaldı mı?:", is_failed)

# Eşitlik Hatası Testi
print(course_grade == "79.0") # Sayı ile String eşit değildir. False döner.
```

### Type Conversion (Casting)
- Python dinamik tipli bir dildir. Değişkenin saklayacağı veri tipini declare etmemiz gerekmez.
- Lakin strongly typed bir dil olduğu için veri tipleri birbirleriyle beraber kullanılamazlar.
- Veri tiplerini birbirine dönüştürme işine casting denir.
    * Implicit (Örtülü): Python'un otomatik yaptığı (Örn: 3 + 4.5 işleminde 3 otomatik olarak 3.0 olur.)
    * Explicit (Açık): Bizim fonksiyon kullanarak kendimizin yaptığı

Kritik Dönüştürücüler: 
1. int()
2. float()
3. str()
4. bool() -> 0 ve boşluklar False, geri kalan her şey True olur.

> Bir sayıyı string'den direkt integer'a çeviremeyiz. int("3.14") ValueError Verir.
>
> Önce float() yapıp sonra int() yapmamız gerekir.

```python
# Hatırlatma: course_grade -> 79.0 "float"

# Float to Integer 
grade_int = int(course_grade)
print(f"Orijinal: {course_grade}, Integer Hali: {grade_int}")

# Int/Float to String (Concatenation için gerekli)
message = "Notunuz: " + str(course_grade)
print(message)

# String to Int/Float
sensor_data = "24.8"
try: 
    sicaklık = int(sensor_data)
except ValueError as hata:
    print("Hata: String doğrudan integer olmaz.")
    print("HATA MESAJI: ", hata)

sicaklik = float(sensor_data)
tam_derece = int(sicaklik)
print(f"Ölçülen: {sicaklik}, Ekranda Görünen: {tam_derece}")

# True/False Test
print(bool(1))
print(bool())
print(bool(""))
print(bool(0))
print(bool("A"))
```
