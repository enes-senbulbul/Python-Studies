## 1. String
- Python'da char tipi yoktur. Stringler uzunluğu 1 olan stringlerden oluşan sequencelerdir.
- Docstring fonksiyon/dosya açıklaması olarak yazılan üç tırnaklı stringlere deniyor.
- Çok satırlı stringleri variable'a atayabiliriz. print ile yazdırdığımızda da çok satırlı halini korur.
- stringler üzerinde for ile loop yapabiliyoruz.
- methodlar:
    - .upper()
    - .lower()
    - .strip(): Baştaki ve sondaki whitespace'i siler.
    - .capitalize(): ilk harfini büyük yapar.
    - .replace("A","B"): stringdeki bir string parçasını diğer string parçası ile değiştirir.
    - .split(): Birden fazla string'e böler ve liste içinde döndürür.
- String Formatting:
    - 3.6 versiyonu ile f-string özelliğinin gelmesiyle .format() kullanma derdinden kurtulduk.
    - placeholder{}'ların içindeki variable'ların value'sini koyar.
    - :.2f, :,(bölüklerin arasına virgül koyar), :_(alt çizgi koyar) vs. value'nun gösterim şeklini de modify edebiliriz.
- Escape characters:
    - \n, \\, \', \" gibi    

````python
# Çok satırlı stringler
a = """Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua."""
print(a)

# String Methodları (Hepsi döndürmelidir. Stringler ummutable olduğu için hiçbiri in-place değildir.)
penguen = "   Penguenleri Seviyorum   "
penguen_upper = penguen.upper()
penguen_lower = penguen.lower()
penguen_strip = penguen.strip()
penguen_capitalize = penguen.capitalize()
penguen_split = penguen.split(" ")
senguen = penguen.replace("P","Ş")
print(penguen, penguen_upper, penguen_lower, penguen_capitalize, penguen_strip, penguen_split, senguen, sep="\n")

# Iterable olduğundan string üzerinde loop yapabiliyoruz.
string_example = "Güvercin"
for letter in string_example:
    print(letter, end=" ")

# Sequence olduğu için indeksleme, slicing vs. vardır. len() kesin çalışır.
string_example_2 = "Herkese Merhaba"
string_sliced_1 = string_example_2[5:]
string_sliced_2 = string_example_2[:5]
string_sliced_3 = string_example_2[-5:-2]
print("\n",string_sliced_1, string_sliced_2, string_sliced_3, sep="\n")

# Sequence olduğu için concatenate ve multiply yapabiliyoruz.
ad = "Enes"
soyad = "Şenbülbül"
ad_soyad = ad + " " + soyad
print(ad_soyad)
print(ad * 5)
print("-"*50)

# f-strings
sayi = 1_000_000
sentence = f"Gün sonunda {sayi} kişi istihdam edildi."
sentence_2 = f"Gün sonunda {sayi:,} kişi istihdam edildi."
sentence_3 = f"Gün sonunda {sayi:_} kişi istihdam edildi."
print(sentence)
print(sentence_2)
print(sentence_3)

price_tl = 59
dolar_to_tl = 45
id_num = 575
print(f"Fiyat {price_tl * dolar_to_tl :.2f} tuttu.")
print(f"Fişte yazan kimlik numarası: {id_num:05d}")

price = 49
txt = f"It is very {'Expensive' if price>50 else 'Cheap'}"
# içinde ayrıca fonksiyon kullanıp dönen değeri de yazdırabiliyoruz. Herhangi bir expression olması yetiyo

string_straight = "Burada {} gün daha kalacağım."
print(string_straight.format(7))
string_straight_2 = "Benim adım {ad}, {yas} yaşındayım. {yas} yıldır yaşıyorum.".format(yas=22, ad="Enes")
print(string_straight_2)
adet = 5
hava_durumu = "Güneşli"
paket = (adet,hava_durumu)
print("Burada %d adet elma var. Hava bugün %s" % paket)
````

## 2. List
- indeksleme, accessing, len vs. var.
- iterable olduğu için üzerinde for ile looping yapabiliyoruz.
- sequence olduğu için + ile join de yapabiliyoruz.
- methodlar:
    - .append(a), insert(i,a), extend(iterable)
    - .remove(a), pop(i), del keyword, clear()
- list comprehension
</br>
```python
      newlist = [expression for item in itarable if condition == True]
      example = [number for number in numbers if number % 2 == 0]
```
</br>

- .sort(), .reverse()
- .copy(), list(), [:] for copying lists

````python
# List Methodları (Hepsi in-place'dir.)
mylist = ["apple", "banana", "cherry"]
myset = {"grapes", "peach", "kiwi"}
mylist.append("watermelon"); print(mylist)
mylist.insert(0, "pineapple"); print(mylist)
mylist.extend(myset); print(mylist)
print("-"*30)
mylist.remove("apple"); print(mylist)
pop_item = mylist.pop(2); print(mylist, pop_item, sep=" | ")
del mylist[2:4]; print(mylist)
mylist.clear(); print(mylist, "clear() methodu listenin içini tamamen boşaitmış")
del mylist; 
try:
    print(mylist)
except NameError:
    print("del ile listeyi tamamen silmişiz.", "\n", "-"*30)

# Iterable olduğundan list üzerinde loop yapabiliyoruz.
mylistagain = ["elma", "armut","kiraz", "kayısı", "ananas"]
for fruit in mylistagain:
    print(fruit, end=" ")

# Sequence olduğu için indeksleme, slicing vs. vardır. len() kesin çalışır.
list_example_2 = ["elma", "armut","kiraz", "kayısı", "ananas"]
list_sliced_1 = list_example_2[2:]
list_sliced_2 = list_example_2[:2]
list_sliced_3 = list_example_2[-4:-2]
print("\n", list_sliced_1, list_sliced_2, list_sliced_3, sep="\n", end="\n"+str('-'*20)+"\n")

# Sequence olduğu için concatenate ve multiply yapabiliyoruz.
list_alpha = ["elma", "armut","kiraz", "kayısı", "ananas"]
list_beta = ["karpuz", "muz", "greyfurt", "portakal", "mandalina"]
list_joined = list_alpha + list_beta
print(list_joined)
print(list_beta * 5)
print("-"*50)

# copying lists
liste1 = ["kırmızı", "mavi", "yeşil", "sarı", "turuncu"]
liste2 = liste1 # yanlış
liste3 = liste1.copy()
liste4 = list(liste1)
liste5 = liste1[:]

# list comprehension
cities = ["istanbul", "yalova", "bursa", "balıkesir", "kocaeli", "tekirdağ"]
new_list = []
for city in cities:
    if "r" in city:
        new_list.append(city) #in-place
"""Ya da şu şekilde yapabiliriz"""
new_list_2 = [city for city in cities if "r" in city]
print(new_list)
print(new_list_2)

# Sorting lists

#customizing sort function
def myFunc(number):
    return abs(number-100)

numbers = [100, 50, 46, 79, 32, 172] 
names = ["Alex", "chris", "Jordan", "taylor", "sam", "Jamie"]
numbers.sort()
names.sort()
print("-"*20,"\nArtan ve Alphanumeric sıralı:", numbers, names, sep="\n")
numbers.sort(reverse = True)
names.sort(reverse = True)
print("-"*20,"\nAzalan ve Alphanumeric -tersten- sıralı:", numbers, names, sep="\n")
numbers.sort(key = myFunc)
names.sort(key = str.upper)
print("-"*20,"\n100'e yakınlık ve Alfabetik -büyük küçük harf farketmez- sıralı:", numbers, names, sep="\n")
numbers.reverse()
names.reverse()
print("-"*20,"\nÖnceki listenin tersine çevrilmiş hali:", numbers, names, sep="\n")
````

## 3. Tuple
- indeksleme, accessing, len vs. var. Ama değişiklik yapabilmemiz için list'e çevirmemiz gerekiyor(workaround).
- iterable olduğu için for ile üzerinde looping yapabiliyoruz.
- list'e çevirdikten sonra işte append, insert, extend, remove, pop vs. kullanabiliyoruz.
- sequence olduğu için + ile join yapabiliyoruz.  ile multiple da yapabiliyoruz. Bunlar için list'e çevirmemize gerek yok.
- list'e çevirmezsek eğer del ile sadece tuple'ın tamamını silebiliyoruz.
- Asteriks kullanımı

````python
# Iterable olduğundan tuple üzerinde loop yapabiliyoruz.
mytuple = ("elma", "armut","kiraz", "kayısı", "ananas")
for fruit in mytuple:
    print(fruit, end=" ")

# Sequence olduğu için indeksleme, slicing vs. vardır. len() kesin çalışır.
tuple_sliced_1 = mytuple[2:]
tuple_sliced_2 = mytuple[:2]
tuple_sliced_3 = mytuple[-4:-2]
print("\n\n",len(mytuple))
print("\n", tuple_sliced_1, tuple_sliced_2, tuple_sliced_3, sep="\n", end="\n"+str('-'*20)+"\n")

# Sequence olduğu için concatenate ve multiply yapabiliyoruz.
tuple_alpha = ("elma", "armut","kiraz", "kayısı", "ananas")
tuple_beta = (1, 5, 6, "portakal", True)
tuple_joined = tuple_alpha + tuple_beta
print(tuple_joined)
print(tuple_beta * 5)
print("-"*50)

# copying lists
tuple1 = ("kırmızı", "mavi", "yeşil", "sarı", "turuncu")
tuple2 = tuple1 # yanlış
# tuple3 = tuple1.copy() # Bu da yanlış
tuple4 = tuple(tuple1)
tuple5 = tuple1[:]

# Packing/Unpacking - Asteriks Kullanımı
tuple_fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
(green, yellow, *red) = tuple_fruits
print(green)
print(yellow)
print(red) # liste olarak döndürür
````

## 4. Dict
- Verileri key:value çiftleri halinde tutuyor.
- Accessing indeksleme ile değil key'ler ile çalışır. Ayrıca .get(key) ile de accessing yapabiliriz ama içeriği değiştiremeyiz. len() de çalışıyor.
    - .keys(): tüm keyleri listeler.
    - .values(): tüm değerleri listeler.
    - .items(): tüm key, value çiftlerini listeler.
- Methodlar
    - .update(dict) 
    - .pop(key), popitem() -> en son ekleneni atar.
    - clear() -> sözlüğü boşaltır.
    - del keyword -> sözlüğü tamamen siler.
- for ile looping yaptığımızda
    - normalde key'ler üzerinden gider.
    - mydict.values ile değerler, mydict.items() ile de keys, values ikilisi üzerinden looping yapabiliriz.
- .copy(), dict() for copying
- nested dictionaries, accessing

````python
# Dictionary Methodları (Neredeyse tamamı in-place'dir.)
my_dict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1969
}; print(my_dict)

my_dict.update({"year":1970, "color":"green"})
my_dict["color"] = "blue"
print(my_dict.get("year", "yıl girilmemiş.")) 
print(my_dict)
pop_item = my_dict.popitem() # En son ekleneni atar. Yine sözlğk olarak döndürür.
print(my_dict, pop_item, sep="  |  ")
my_dict.clear()
del my_dict


# Iterable olduğu için dictionary üzerinde loop yapabiliyoruz.
student_dict = {
    "name": "Enes Penguen",
    "gender": "male",
    "age": 22,
    "department": "MatMüh",
    "student": True
} 

print("\n", "-"*20)
for anahtar in student_dict:
    print(anahtar, end=" ")

print("\n", "-"*20)
for anahtar in student_dict:
    print(student_dict[anahtar], end=" ")

print("\n", "-"*20)
for anahtar in student_dict.keys():
    print(anahtar, end=" ")

print("\n", "-"*20)
for deger in student_dict.values():
    print(deger, end=" ")

print("\n", "-"*20)
for anahtar, deger in student_dict.items():
    print(anahtar.upper() + ": " + str(deger))


# Copying dicts
print("\n", "-"*20)
student_2 = dict(student_dict)
student_3 = student_dict.copy()
print(student_2)
print(student_3)


# Nested Dictionaries
print("\n", "-"*20)

dict_child1 = {
    "age": 19,
    "gender": "male",
    "education": "university"
}

dict_child2 = {
    "age": 14,
    "gender": "female",
    "education": "highschool"
}

my_family = {
    "child1": dict_child1,
    "child2": dict_child2
}

print(my_family["child1"]["age"])

for anahtar, deger1 in my_family.items():
    for anahtar2 in deger1:
        print(anahtar, anahtar2, deger1[anahtar2], sep=": ")


# dict constructur dict()
print("\n", "-"*20)
thisdict4 = dict(name= "John", age = 36,country = "Norway") #DICT CONSTRUCTOR
print(thisdict4)
````

## 5. Set
- Aynı elemandan iki tane bulunamaz. False ile 0'ın değeri(value) aynı olduğundan ikisini tek bir eleman sayar.
- len() fonksiyonu çalışır.
- for ile looping yapabiliyoruz ama elemanları rastgele seçiyor.
- Methodlar:
    - .add(a), .update(iterable)
    - .remove(a), discard(a), pop() -> random
    - clear() -> kümeyi boşaltır. boş küme döndürür.
    - del -> kümeyi tamamen siler.
- Join sets -> + ve * çalışmaz çünkü sequence değil. Onun yerine |, &, ^, -(difference) var.
    - .union(set), .update(any itetable)
    - .intersection()
    - .difference()
    - .symmetric_difference()
    - bu yukarıdakiler normalde yeni küme döndürüyorlar ama "_update" suffix'ini almış halleri in-place çalışıp kümenin kendisini değiştiriyır.

````python
# set methodları (bazıları in-place, bazıları değer döndüren türdendir)
my_set = {"defter", "telefon", "kronometre", "monitör", "laptop"}; print(my_set)
tuple_things = ("tabanlık", "lamba")
my_set.add("bardak"); print(my_set)
my_set.update(tuple_things); print(my_set)
my_set.remove("telefon") # küme içinde yoksa hata verir.
my_set.discard("telefon"); print(my_set) # küme içinde yoksa hata vermez.
item_pop = my_set.pop(); print(my_set, item_pop, sep="   |   ")
my_set.clear(); print(my_set, "-> Boş küme",end="\n"+str('-'*20)+"\n")
del my_set # kümenin tamamını siler

# iterable olduğundan üzerinde loop yapabiliyoruz ama RANDOM seçiyor.
my_set_again = {"defter", "telefon", "kronometre", "monitör", "laptop"}
for birsey in my_set_again:
    print(birsey, end=" ")

# Sequence değil. Üzerinde başka operatörlerle joining yapabiliyoruz. Multiply yok.
set_alpha = {"elma", "armut","kiraz", "kayısı", "ananas"}
set_beta = {"elma","kiraz", "ananas", "portakal", "mandalina"}
set_joined = set_alpha | set_beta
print("-"*50)
````
