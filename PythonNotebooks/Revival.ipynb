# Adım 6: Kümeler ve Veri Temizliği (Deduplication)

# Adım 7: Sağlam (Robust) Veri Grişi Döngüsü

```python
while True:
    stok_adeti = input("Stok adeti giriniz:\n")
    try:
        stok_adeti = int(float(stok_adeti))
    except ValueError as hata_mesaji:
        print(hata_mesaji, "\nLütfen sayısal değer giriniz.")
        continue
    if stok_adeti <= 0:
        print("Stok adeti pozitif olmalı. Senin girdiğin değer: %d" % stok_adeti)
        continue
    else:
        break

print("Geçerli değer girdiniz: {adet}".format(adet=stok_adeti))
```

# Adım 8: Fonksiyonlar, Parametreler ve Return

```python
def tl_cevir(dolar_fiyati, dolar_kuru = 30):
    tl_fiyati = dolar_fiyati * dolar_kuru
    return tl_fiyati

dolar_fiyat = 100
kriz_kuru = 40
print("Standart Kur ile:", tl_cevir(dolar_fiyat))
print("Yüksek Kur ile:", tl_cevir(dolar_fiyat, kriz_kuru))
```

# Adım 9. Çoklu Return ve Tuple Unpacking

```python
def daire_hesapla(yaricap, pi = 3.14):
    alan = pi * (yaricap**2)
    cevre = 2 * pi * yaricap
    return alan, cevre

val_yaricap = int(float(input("Yarıçap kaç birim?\n")))
alan, cevre = daire_hesapla(yaricap=val_yaricap)
print(f"Alan: {alan} Çevre: {cevre}")
```

# Adım 10: The Aliasing Trap, Identity Operators

```python
gorevler = list(("Rapor Yaz", "Mail At"))
yedek_yanlis = gorevler
yedek_dogru = list(gorevler)
gorevler.extend({"Toplantı Yap"})
print("Yedek Yanlış:", yedek_yanlis, "\nYedek Doğru", yedek_dogru)
print("gorevler ile yedek_yanlis aynı nesne mi:", gorevler is yedek_yanlis)
print("gorevler ile yedek_dogru aynı nesne mi:", gorevler is yedek_dogru)
```

# Adım 12: Final Challange

```python
import random

# --- FONKSİYONLAR ---
def bakiye_sorgula(hesap_no_given, database):
    print("\n>> Bakiye Sorgulanıyor...")
    valid_isim = database[hesap_no_given]["ad"]
    valid_bakiye = database[hesap_no_given]["bakiye"]
    print(f"Sayın {valid_isim}\nHesabınızdaki bakiye: {valid_bakiye} TL")

def para_yatir(hesap_no_given, database):
    while True:
        try:
            miktar_yatir = int(float(input("Yatırmak istediğiniz miktarı giriniz.")))
        except ValueError:
            print("(!) Lütfen Sayısal Değer Giriniz.")
            continue
        if miktar_yatir <= 0:
            print("(!) Lütfen pozitif değer giriniz.")
            continue
        else:
            print(">> Para yatırma başarılı...")
            print(f"Yatırılan miktar: {miktar_yatir} TL")
        break
    return miktar_yatir
                        
def para_cek(hesap_no_given, database):
    guncel_bakiye = database[hesap_no_given]["bakiye"]
    print(f"Hesabınızdaki güncel bakiye: {guncel_bakiye} TL")
    while True:
        try:
            miktar_cek = int(float(input("Çekmek istediğiniz miktarı giriniz.")))
        except ValueError:
            print("(!) Lütfen Sayısal Değer Giriniz.")
            continue
            
        if miktar_cek <= 0:
            print("(!) Lütfen pozitif değer giriniz.")
            continue
        if miktar_cek > guncel_bakiye:
            print("(!) Hesabınızda yeterli para yok. Yeni bir miktar girin.")
            continue
        else:
            print(">> Para çekme başarılı...")
            print(f"Çekilen miktar: {miktar_cek} TL")
        break
    return miktar_cek

def hesap_ozeti_goster(hesap_no_given, database):
    print("\n--- Hesap Özeti ---")
    print(f"Müşteri Hesap No {hesap_no_given}")
    print(f"Müşteri: {database[hesap_no_given]["ad"]}")
    print("\n~ İşlem Geçmişi ~")
    for islem in database[hesap_no_given]["gecmis"]:
        print("---",islem)


# --- VERİ TABANI OLUŞTURMA ---
names = [
    "Alex", "Emma", "Lucas", "Olivia", "Noah",
    "Sophia", "Ethan", "Mia", "Liam", "Ava",
    "Daniel", "Isabella", "James", "Charlotte",
    "Benjamin", "Amelia", "Henry", "Emily",
    "Michael", "Grace"
]

islemler = {
    1: "Bakiye Sorgula",
    2: "Para Yatır",
    3: "Para Çek",
    4: "Hesap Özeti",
    5: "Çıkış Yap"
}

database = dict()

# Sabit Test Kullanıcısı
database["12345"] = {"pin": "0000", "bakiye": 5000, "ad": "Enes", "gecmis": []}

# Rastgele Kullanıcılar
for t in range(9):
    hesap_no = str(12346+t)
    if hesap_no in database.keys():
        continue
    pin = f"{int(random.randint(0,9999)):04d}"
    bakiye = random.randint(1,10) * 1000
    ad = random.choice(names)
    gecmis = list()
    dict_person = dict(pin=pin, bakiye=bakiye, ad=ad, gecmis= [])
    database[hesap_no] = dict_person

# --- GİRİŞ EKRANI (LOGİN) ---
print("**************** BANKAMATİK SİSTEMİNE HOŞGELDİNİZ ****************")
while True:
    try:
        giris = input("\nHesap no giriniz (Çıkış için 'q'): ")
        if giris == 'q': exit()
        hesap_no_given = str(int(float(giris))) # Float güvenliği
    except ValueError:
        print("(!) Lütfen sayısal bir değer giriniz...")
        continue
    hesap_no_given = str(hesap_no_given)
    if hesap_no_given not in database.keys():
        print("(!) Lütfen Geçerli Bir Hesap Numarası Giriniz:")
        del hesap_no_given
        continue
    else:
        break
    
valid_pin = database[hesap_no_given]["pin"]
valid_pin_numeric = int(valid_pin)

k = 0
while True:
    try:
        pin_given = int(float(input("Lütfen pin kodunu giriniz.")))
    except ValueError:
        print("Lütfen sayısal bir değer giriniz...")
        continue
    if pin_given != valid_pin_numeric:
        print("Yanlış PIN girdiniz.")
        k += 1
        if k <= 3:
            print(f"Kalan Hakkınız: {3-k}")
        else:
            print("Tüm haklarınız bitti. Kartınız bloke oldu!!!")
            exit()
        continue
    else:
        print("PIN onaylandı. Giriş Yapılıyor")
        break

# --- ANA MENÜ ---
while True:
    
    valid_isim = database[hesap_no_given]["ad"]
    
    while True:
        print(f"\nMerhaba {valid_isim.upper()},\n--- Menü ---")
        for m in range(5):
            print(f"{m+1}. {islemler[m+1]}")
        islem = int(input("Yapmak istediğiniz İşlemi Seçiniz."))
        if islem not in islemler.keys():
            print("Lütfen geçerli bir işlem seçin.")
            continue
        else:
            break
    
    islem_adi = islemler[islem] 
    
    match islem:
        case 1:
            bakiye_sorgula(hesap_no_given, database)
            database[hesap_no_given]["gecmis"].append(islem_adi)
        case 2:
            # DÜZELTME 2: Global bakiye yerine database'den güncelleme
            miktar = para_yatir(hesap_no_given, database)
            database[hesap_no_given]["bakiye"] += miktar 
            
            guncel = database[hesap_no_given]["bakiye"]
            print(f"Güncel Hesap Bakiyesi: {guncel} TL")
            database[hesap_no_given]["gecmis"].append(f"{islem_adi}: +{miktar} TL")
        case 3:
            # DÜZELTME 2: Global bakiye yerine database'den güncelleme
            miktar = para_cek(hesap_no_given, database)
            database[hesap_no_given]["bakiye"] -= miktar
            
            guncel = database[hesap_no_given]["bakiye"]
            print(f"Güncel Hesap Bakiyesi: {guncel} TL")
            database[hesap_no_given]["gecmis"].append(f"{islem_adi}: -{miktar} TL")
        case 4:
            hesap_ozeti_goster(hesap_no_given, database)
            database[hesap_no_given]["gecmis"].append(islem_adi)
        case 5:
            print("Programdan çıkış Yapılıyor. İyi günler!")
            exit()
            
    devam = input("\nBaşka İşlem Yapmak İster Misiniz? (Evet için 'e', Hayır için herhangi bir tuş): ")
    if devam.lower() == "e" or devam.lower() == "evet":
        continue
    else:
        print("Çıkış yapılıyor...")
        break

exit()
```
