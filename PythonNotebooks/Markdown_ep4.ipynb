### Görev Listeleri (Task Lists)
Standart madde işaretlerinin işaretlenebilir kutucuklara (checkbox) dönüştürülmüş halidir. Özellikle GitHub issues, Pull Request açıklamaları veya kişisel .md dosyalarında proje ilerlemesini takip etmek için kullanılır.

Syntax: sırasız liste mantığıyla (- veya *) yapılır ama yanına bir de köşeli parantez eklenir. Sıra işaretinden sonra ve köşeli parantezlerin içinde birer boşluk olmak zorundadır.

- [x] Repo oluşturuldu.
- [x] README.md dosyası eklendi.
- [ ] Python sanal ortamı (venv) kurulacak
- [ ] Requirements.txt hazırlanacak.

Normalde kendimiz listeyi oluştururken içine X koymak zorundayız ama GitHub ve GitLab üzerinde bu listeyi oluşturduğumuzda öngörüntüleme (preview) modunda veya dosya kaydedildikten sonra kutucuklara tıklayabiliriz.

### HTML Entegrasyonu
Markdown syntax'ının yetersiz kaldığı durumlarda (örn: metni ortalamak, resim boyutlandırmak veya renkli yazı yazmak) dokümanın içine doğrudan html kodları yazabiliriz.

Markdown'da ortala diye bir komut yoktur. Bunun için div veya p etiketi kullanılır. 

#### Sık Kullanılan HTML Hileleri
##### Metni veya görseli ortalamak
<div align="center">
    <img src="logo.png" width="100">
    <h3>Bu başlık ve logo ortalandı.</h3>
</div>

##### Katlanabilir Alanlar (Collapsible Sections)
- Mühendislik dökümanlarında uzun log kayıtlarını veya spoiler içerikleri gizlemek için harikadır.

<details>
    <summary>Tıklayarak Detayları Gör...</summary>

    Buraya çok uzun hata kodları,
    loglar veya ekstra açıklamalar gelebilir.
    Markdown syntax'ı html bloğu içinde de çalışır.
    - Madde 1
    - Madde 2
</details>

##### Satır Atlatma
- Bazen Markdown'da paragraf boşlukları çok geniş gelir veya tablo içinde alt satıra geçmek istersiniz.

Satır 1<br>Satır 2(Hemen Altında)

* GitHub ve çoğu platfor mgüvenlik sebebiyle <script>, <iframe> veya karmaşık style="" etiketlerini temizler. Ancak <img>, <div>, <table>, <details> gibi yapısal etiketlemelere izin verir.
* HTML etiketlerinin içine Markdown yazarken bazen boş satır bırakmamız gerekir. Özellikle <details> etiketinin içindeki içeriğin Markdown olarak render edilmesi için üstte ve altta birer boşluk bırakmalıyız.
