## Satır İçi Kod ve Kod Blokları 
- Yazdığın metnin içinde "Bu bir yazı değil, bu çalıştırılabilir bir kod parçasıdır." mesajını vermek için kullanılır.
- Markdown'da kod yazmak için ters tırnak (backtick) işareti kullanılır. Bu bizim bildiğimiz normal tırmak (') değildir.
- Kısayolu [Alt] + [ ; ] tuş kombinasyonudur.

#### Satır İçi Kod (Inline Code)
syntax -> `kod`

Python'da ekrana yazdırmak için `print()` fonksiyonu kullanılır. 
Lütfen `main.py` dosyasını açın.

#### Kod Blokları (Code Blocks)
Uzun kod parçalarını, algoritmaları göstermek için kullanılır. Üç adet ters tırnak ile başlar ve biter.

syntax -> 
``` python
def merhaba();
    print("Kod buraya yazılır.")
```
- Üç tırnaktan sonra dilin adını küçük harflerle yazarsak markdown o dile uygun renklendirme yapar.
- Bu okunabilirliği %100 arttırır.

```c
#include <stdio.h>
int main(){
    printf("Bu bir kod bloğudur.");
    return 0;
}
```

# Pisagor Teoremi ve Hipotenüs Hesaplama
Bu teorem ile dik üçgenin iki dik kenarının uzunluklarını kullanarak hipotenüs'ün uzunluğunu hesaplayabiliriz.

Bu hesaplamayı yapmak için şu formülü kullanırız:
$$
a^{2} + b^{2} = c^{2}
$$

Bu hesaplamayı python dilinde de şu şekilde koda dökebiliriz:

> Kenar uzunlukları negatif olamaz.

```python
import math

a = float(input("Dil üçgenin kenarlarından birinin uzunluğunu giriniz."))
b = float(input("Dil üçgenin kenarlarından diğerinin uzunluğunu giriniz."))

c = math.sqrt(a**2 + b**2)
print(f"Dik üçgenin hipotenüsünün uzunluğu {c:.2f} birimdir.")

```
